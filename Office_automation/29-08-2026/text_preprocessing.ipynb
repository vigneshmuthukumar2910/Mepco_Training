```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')

# 1. Paragraph Sample Text Corpus (Technology & History Topic)
text = """Artificial Intelligence has transformed modern industries over the last decade. 
Scientists are developing smarter algorithms that learn from massive datasets automatically. 
While these technologies offer incredible efficiency, researchers are still studying their long-term societal impacts."""

print(f"Original Paragraph:\n{text}\n")

# 2. Tokenization
tokens = word_tokenize(text)
print(f"1. Tokens:\n{tokens}\n")

# 3. Stop-Word Removal (Filtering out 'has', 'over', 'the', 'are', etc.)
stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w.lower() not in stop_words and w.isalnum()]
print(f"2. Filtered Tokens (No Stop-Words):\n{filtered_tokens}\n")

# 4. Stemming
ps = PorterStemmer()
stemmed_tokens = [ps.stem(w) for w in filtered_tokens]
print(f"3. Stemmed Tokens:\n{stemmed_tokens}")
```
