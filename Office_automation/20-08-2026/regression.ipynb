```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

X_b = np.c_[np.ones((len(X), 1)), X]
#theta = (X^T * X)^(-1) * X^T * y
theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

print("Learned Parameters (Intercept, Slope):")
print(theta_best.ravel())
      

y_predict = X_b.dot(theta_best)

plt.figure(figsize=(7, 4))
plt.scatter(X, y, color="blue", alpha=0.6, label="Data points")
plt.plot(X, y_predict, color="red", linewidth=2, label="Linear Regression Fit")
plt.title("Linear Regression from Scratch using NumPy")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()

X_mean = np.mean(X)
X_std = np.std(X)
X_scaled = (X - X_mean) / X_std

X_b_scaled = np.c_[np.ones((len(X_scaled), 1)), X_scaled]
theta_best_scaled = np.linalg.inv(X_b_scaled.T.dot(X_b_scaled)).dot(X_b_scaled.T).dot(y)
y_predict_scaled = X_b_scaled.dot(theta_best_scaled)


mse = np.mean((y - y_predict_scaled) ** 2)
ss_res = np.sum((y - y_predict_scaled) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - (ss_res / ss_tot)

print("\nMetrics after StandardScaler:")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R2 Score: {r2:.4f}")
print(f"Scaled Parameters (Intercept, Slope): {theta_best_scaled.ravel()}")
```
