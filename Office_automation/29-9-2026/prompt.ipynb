```python
# ============================================================
# PROMPT ENGINEERING EXPERIMENT
# Zero-Shot, Few-Shot and Chain-of-Thought
# Using Google Gemini API
# ============================================================

from google import genai
import time
import getpass


# ============================================================
# 1. GEMINI API SETUP
# ============================================================

print("=" * 70)
print("GEMINI API SETUP")
print("=" * 70)

# Enter your NEW Gemini API key securely
api_key = getpass.getpass("")

client = genai.Client(api_key=api_key)

# Stable Gemini model
MODEL_NAME = "gemini-3.5-flash"


# ============================================================
# 2. FUNCTION TO CALL GEMINI
# ============================================================

def ask(prompt):

    # Try up to 3 times in case of temporary 503 errors
    for attempt in range(3):

        try:

            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt
            )

            return response.text

        except Exception as e:

            print(f"\nAttempt {attempt + 1} failed.")
            print("Error:", e)

            # Retry if this was not the final attempt
            if attempt < 2:
                print("Retrying in 5 seconds...\n")
                time.sleep(5)

            else:
                return "API request failed after 3 attempts."


# ============================================================
# 3. ZERO-SHOT PROMPTING
# ============================================================

zero_shot = """
You are performing a sentiment analysis task.

Classify the following sentence as Positive, Negative, or Neutral.

Sentence:
"The new smartphone has an excellent camera and very smooth performance."

Give a detailed response containing:

1. The predicted sentiment
2. The important words that influenced the prediction
3. A short explanation of why the sentence belongs to that category
"""


print("\n" + "=" * 70)
print("1. ZERO-SHOT PROMPTING")
print("=" * 70)

print("\nPROMPT:")
print(zero_shot)

print("\nMODEL RESPONSE:")

zero_result = ask(zero_shot)

print(zero_result)


# ============================================================
# 4. FEW-SHOT PROMPTING
# ============================================================

few_shot = """
You are performing sentiment analysis.

Use the examples below to understand the classification task.

Example 1:
Sentence: "I absolutely love this phone. It is fast and reliable."
Answer: Positive
Explanation: The words "love", "fast", and "reliable" indicate
a positive opinion.

Example 2:
Sentence: "The laptop is slow and frequently crashes."
Answer: Negative
Explanation: The words "slow" and "crashes" indicate a negative
experience.

Example 3:
Sentence: "The package arrived yesterday."
Answer: Neutral
Explanation: This sentence only gives factual information and
does not express a positive or negative opinion.

Now analyze the following sentence:

"The new smartphone has an excellent camera and very smooth performance."

Give a detailed response containing:

1. The predicted sentiment
2. The important words
3. An explanation based on the examples
4. A final conclusion
"""


print("\n" + "=" * 70)
print("2. FEW-SHOT PROMPTING")
print("=" * 70)

print("\nPROMPT:")
print(few_shot)

print("\nMODEL RESPONSE:")

few_result = ask(few_shot)

print(few_result)


# ============================================================
# 5. CHAIN-OF-THOUGHT PROMPTING
# ============================================================

chain_of_thought = """
You are solving a mathematical reasoning problem.

Problem:

A student has 50 notebooks.
He gives 12 notebooks to his friend.
Then he buys 8 more notebooks.
Finally, he gives 10 notebooks to another student.

Determine how many notebooks the student has now.

Solve the problem step by step.
Show the calculations for each operation.

At the end, clearly state the final answer.
"""


print("\n" + "=" * 70)
print("3. CHAIN-OF-THOUGHT PROMPTING")
print("=" * 70)

print("\nPROMPT:")
print(chain_of_thought)

print("\nMODEL RESPONSE:")

cot_result = ask(chain_of_thought)

print(cot_result)


# ============================================================
# 6. FINAL COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("PROMPT ENGINEERING EXPERIMENT COMPLETED")
print("=" * 70)

print("""
Technique 1: Zero-Shot
- No examples were provided.
- The model directly performed the requested task.

Technique 2: Few-Shot
- Examples were provided before the actual question.
- The model learned the expected classification pattern from examples.

Technique 3: Chain-of-Thought
- The model was asked to solve a multi-step reasoning problem.
- The model was instructed to work through the calculations step by step.
""")
```
